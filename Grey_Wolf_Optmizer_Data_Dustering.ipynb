```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score
)

# =====================================================
# 1. LOAD DATASET
# =====================================================

iris = load_iris()
X = iris.data
y_true = iris.target

scaler = StandardScaler()
X = scaler.fit_transform(X)

N = X.shape[0]          # Number of data points
D = X.shape[1]          # Number of features
K = 3                   # Number of clusters

# =====================================================
# 2. KCGWO PARAMETERS
# =====================================================

WOLVES = 30
MAX_ITER = 100

LOWER = X.min(axis=0)
UPPER = X.max(axis=0)

# Each wolf represents K cluster centers
DIM = K * D

# =====================================================
# 3. FITNESS FUNCTION
# =====================================================

def fitness(position):

    centers = position.reshape(K, D)

    distances = np.linalg.norm(
        X[:, np.newaxis, :] - centers[np.newaxis, :, :],
        axis=2
    )

    labels = np.argmin(distances, axis=1)

    # Sum of squared intra-cluster distances
    total = 0.0

    for j in range(K):
        cluster_points = X[labels == j]

        if len(cluster_points) == 0:
            return np.inf

        total += np.sum(
            (cluster_points - centers[j]) ** 2
        )

    return total


# =====================================================
# 4. K-MEANS REFINEMENT
# =====================================================

def refine_position(position):

    centers = position.reshape(K, D)

    distances = np.linalg.norm(
        X[:, np.newaxis, :] - centers[np.newaxis, :, :],
        axis=2
    )

    labels = np.argmin(distances, axis=1)

    new_centers = centers.copy()

    for j in range(K):

        points = X[labels == j]

        if len(points) > 0:
            new_centers[j] = np.mean(points, axis=0)

        else:
            new_centers[j] = X[
                np.random.randint(N)
            ]

    return new_centers.flatten()


# =====================================================
# 5. INITIALIZE WOLF POPULATION
# =====================================================

population = np.random.uniform(
    LOWER,
    UPPER,
    size=(WOLVES, K, D)
)

population = population.reshape(WOLVES, DIM)

# Fitness values
scores = np.array([
    fitness(wolf) for wolf in population
])

convergence = []

# =====================================================
# 6. KCGWO MAIN ALGORITHM
# =====================================================

for iteration in range(MAX_ITER):

    # Sort wolves based on fitness
    sorted_indices = np.argsort(scores)

    alpha = population[sorted_indices[0]].copy()
    beta = population[sorted_indices[1]].copy()
    delta = population[sorted_indices[2]].copy()

    # Linearly decreasing parameter
    a = 2 - 2 * iteration / MAX_ITER

    # -------------------------------------------------
    # K-means grouping of wolf population
    # -------------------------------------------------

    random_value = np.random.rand()

    if random_value > 0.5:

        number_groups = min(3, WOLVES)

        km = KMeans(
            n_clusters=number_groups,
            random_state=42,
            n_init=10
        )

        group_labels = km.fit_predict(population)

        group_best = []

        for group in range(number_groups):

            members = np.where(
                group_labels == group
            )[0]

            if len(members) > 0:

                best_member = members[
                    np.argmin(scores[members])
                ]

                group_best.append(
                    population[best_member].copy()
                )

        # Use the best three available group leaders
        leaders = np.array(group_best)

        leader_scores = np.array([
            fitness(leader) for leader in leaders
        ])

        order = np.argsort(leader_scores)

        alpha = leaders[order[0]]

        if len(order) > 1:
            beta = leaders[order[1]]
        else:
            beta = alpha.copy()

        if len(order) > 2:
            delta = leaders[order[2]]
        else:
            delta = beta.copy()

    # -------------------------------------------------
    # Update wolf positions
    # -------------------------------------------------

    new_population = population.copy()

    for i in range(WOLVES):

        r1 = np.random.rand(DIM)
        r2 = np.random.rand(DIM)

        A1 = 2 * a * r1 - a
        C1 = 2 * r2

        D_alpha = np.abs(C1 * alpha - population[i])
        X1 = alpha - A1 * D_alpha

        r1 = np.random.rand(DIM)
        r2 = np.random.rand(DIM)

        A2 = 2 * a * r1 - a
        C2 = 2 * r2

        D_beta = np.abs(C2 * beta - population[i])
        X2 = beta - A2 * D_beta

        r1 = np.random.rand(DIM)
        r2 = np.random.rand(DIM)

        A3 = 2 * a * r1 - a
        C3 = 2 * r2

        D_delta = np.abs(C3 * delta - population[i])
        X3 = delta - A3 * D_delta

        # Paper's weighted position update
        new_position = (
            3 * X1 + 2 * X2 + X3
        ) / 6.0

        # Keep centers inside dataset bounds
        new_position = np.clip(
            new_position,
            np.tile(LOWER, K),
            np.tile(UPPER, K)
        )

        # K-means-inspired centroid refinement
        new_position = refine_position(new_position)

        new_population[i] = new_position

    # -------------------------------------------------
    # Evaluate new population
    # -------------------------------------------------

    new_scores = np.array([
        fitness(wolf) for wolf in new_population
    ])

    # Greedy selection
    for i in range(WOLVES):

        if new_scores[i] < scores[i]:

            population[i] = new_population[i]
            scores[i] = new_scores[i]

    # Record best solution
    best_index = np.argmin(scores)

    convergence.append(scores[best_index])

    print(
        f"Iteration {iteration + 1}/{MAX_ITER}, "
        f"Best Fitness = {scores[best_index]:.4f}"
    )


# =====================================================
# 7. FINAL CLUSTER CENTERS
# =====================================================

best_index = np.argmin(scores)

best_centers = population[
    best_index
].reshape(K, D)

# Assign each data point to nearest center
distances = np.linalg.norm(
    X[:, np.newaxis, :] -
    best_centers[np.newaxis, :, :],
    axis=2
)

final_labels = np.argmin(distances, axis=1)

# =====================================================
# 8. PERFORMANCE EVALUATION
# =====================================================

print("\n========== KCGWO RESULTS ==========")

print("Best Fitness:", scores[best_index])

print(
    "Silhouette Score:",
    silhouette_score(X, final_labels)
)

print(
    "Davies-Bouldin Index:",
    davies_bouldin_score(X, final_labels)
)

print(
    "Calinski-Harabasz Index:",
    calinski_harabasz_score(X, final_labels)
)

print(
    "Adjusted Rand Index:",
    adjusted_rand_score(y_true, final_labels)
)

# =====================================================
# 9. CONVERGENCE GRAPH
# =====================================================

plt.figure(figsize=(8, 5))

plt.plot(
    convergence,
    color="blue",
    linewidth=2
)

plt.xlabel("Iterations")
plt.ylabel("Fitness (SSE)")
plt.title("KCGWO Convergence Curve")
plt.grid(True)

plt.show()

# =====================================================
# 10. CLUSTER VISUALIZATION
# =====================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=final_labels,
    cmap="viridis",
    s=40
)

plt.scatter(
    best_centers[:, 0],
    best_centers[:, 1],
    color="red",
    marker="X",
    s=200,
    label="Cluster Centers"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("KCGWO Data Clustering")
plt.legend()
plt.show()

# =====================================================
# 11. SAVE CLUSTERED DATA
# =====================================================

result = pd.DataFrame(
    X,
    columns=iris.feature_names
)

result["Cluster"] = final_labels

result.to_csv(
    "KCGWO_Clustering_Results.csv",
    index=False
)

print("\nClustering results saved successfully!")
```
